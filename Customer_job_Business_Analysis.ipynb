```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.info()
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
lables=['Young Adult','Adult','Middle-Aged','Senior Citizen']
df['age_group']=pd.qcut(df['age'],q=4,labels=lables)
```

```python
df[['age','age_group']].head()
```

```python
df['frequency_of_purchases'].unique()
```

```python
freq_map={
    'Fortnightly':14,
    'Weekly':7,
    'Annually':365,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Monthly':30,
    'Every 3 Months':90
}
```

```python
df['frequency_of_purchases(in_days)']=df['frequency_of_purchases'].map(freq_map)
```

```python
df.drop('promo_code_used',axis=1)
```
